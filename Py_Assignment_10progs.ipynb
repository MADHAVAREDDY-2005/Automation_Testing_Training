1.	Print all prime numbers between input range (Ex – input 20 50, prints all prime numbers between 20 and 50).

```python
start,end=map(int,input().split())
for num in range(start,end+1):
  prime=True
  for i in range(2,num):
    if num%i==0:
      prime=False
  if prime:
    print(num)
```

2.	Factorial using recursion

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  return n*factorial(n-1)

print(factorial(5))
```

3.	Square of numbers using lambda

```python
lst=list(map(int, input().split()))
key=lambda x: x*x
res=[]
for num in lst:
  res.append(key(num))
print(res)

# result = list(map(lambda x: x * x, lst))
```

4.	Find the second largest element in a list

```python
nums=list(map(int, input().split()))
larg=nums[0]
slarg=nums[0]
for num in nums:
  if num>larg:
    slarg=larg
    larg=num
  elif num>slarg and num<larg:
    slarg=num

print(slarg)
```

5.	Count frequency of characters in a string

```python
string=input()
freq={}
for i in string:
  if i not in freq:
    freq[i]=1
  else:
    freq[i]+=1

print(freq)
```

6.	Calculate area of a circle using math library.

```python
import math
def area_of_circle(radius):
  return math.pi*radius*radius
print(area_of_circle(5))
```

7.	Reverse a string without using built‑in reverse

```python
def rev_str(string):
  res=str()
  for i in range(len(string)-1,-1,-1):
    res+=string[i]
  return res

# print(rev_str("madhav"))
input=input()
print(rev_str(input))
```

8.	Remove duplicates from a list

```python
nums=list(map(int,input().split()))
res=[]
for num in nums:
  if num not in res:
    res.append(num)

print(res)
```

9.	Merge two dictionaries

```python
dict1={'name': 'Madhav', 'age': 21, 'city': 'Chennai'}
dict2={'course': 'AIML', 'year': 4, 'cgpa': 8.4}
for key,value in dict2.items():
  dict1[key]=value
print(dict1)
```

10.	Fibonacci series using recursion

```python
def fibonacci(n):
  if n==0:
    return 0
  if n==1:
    return 1
  return fibonacci(n-1)+fibonacci(n-2)


n=int(input())
for i in range(n):
  print(fibonacci(i),end=" ")
```
